# Unidade 2 – Memória e Dados
### Sistema de Cadastro para Biblioteca
Este programa recebe dados textuais e numéricos de um livro, converte a string de disponibilidade em um valor booleano (`True` ou `False`) e exibe o cadastro formatado.

```python
# Solicita as informações do livro ao usuário
titulo_livro = input("Digite o título do livro: ")
autor = input("Digite o autor: ")
ano_publicacao = int(input("Digite o ano de publicação: "))
disponivel = input("O livro está disponível? (sim/não): ")

# Converte a resposta textual em um valor Booleano (True/False)
disponivel = True if disponivel.lower() == "sim" else False

# Exibe o relatório do cadastro realizado
print("\nCadastro realizado com sucesso!")
print(f"Título: {titulo_livro}")
print(f"Autor: {autor}")
print(f"Ano: {ano_publicacao}")
print(f"Disponível: {disponivel}")
```

# Unidade 3 – Operadores e Expressões
### Calculadora de IMC
Este script calcula o Índice de Massa Corporal (IMC) utilizando operadores aritméticos de potenciação e divisão, classificando o resultado por meio de condicionais encadeadas.

```python
# Coleta os dados de peso e altura convertendo para ponto flutuante (float)
peso = float(input("Digite seu peso (kg): "))
altura = float(input("Digite sua altura (m): "))

# Utiliza a função calcular_imc para obter o IMC e a classificação
imc, classificacao = calcular_imc(peso, altura)

# Exibe o IMC formatado com duas casas decimais e a classificação
print(f"Seu IMC é: {imc:.2f}")
print(f"Classificação: {classificacao}")
```

### Função para Cálculo de IMC
Vamos refatorar o código de cálculo do IMC em uma função para promover a reutilização.

```python
def calcular_imc(peso, altura):
    """
    Calcula o Índice de Massa Corporal (IMC) e retorna a classificação.

    Args:
        peso (float): O peso da pessoa em quilogramas (kg).
        altura (float): A altura da pessoa em metros (m).

    Returns:
        tuple: Uma tupla contendo o valor do IMC (float) e a classificação (str).
    """
    imc = peso / (altura ** 2)

    if imc < 18.5:
        classificacao = "Abaixo do peso"
    elif imc < 25:
        classificacao = "Peso normal"
    elif imc < 30:
        classificacao = "Sobrepeso"
    else:
        classificacao = "Obesidade"

    return imc, classificacao

# Exemplo de uso da função:
# imc_valor, imc_classificacao = calcular_imc(70, 1.75)
# print(f"Seu IMC é: {imc_valor:.2f}")
# print(f"Classificação: {imc_classificacao}")
```

# Unidade 4 – Tomada de Decisão I
### Verificação de Aprovação de Aluno
Cálculo da média aritmética simples de três notas e determinação do status acadêmico do aluno (Aprovado, Recuperação ou Reprovado).

```python
# Entrada das três notas parciais do estudante
nota1 = float(input("Digite a primeira nota: "))
nota2 = float(input("Digite a segunda nota: "))
nota3 = float(input("Digite a terceira nota: "))

# Calcula a média aritmética simples
media = (nota1 + nota2 + nota3) / 3
print(f"Média final: {media:.2f}")

# Aplicação das regras de negócio para validação do status
if media >= 7:
    print("Aluno Aprovado")
elif media >= 5:
    print("Aluno em Recuperação")
else:
    print("Aluno Reprovado")
```

# Unidade 5 – Tomada de Decisão II
### Classificação de Atletas por Idade
Utilização de uma estrutura condicional mais robusta e ramificada para segmentar atletas em categorias de acordo com a faixa etária informada.

```python
# Recebe a idade do atleta como inteiro
idade = int(input("Digite a idade do atleta: "))

# Estrutura sequencial de validação de faixas etárias
if idade < 12:
    print("Categoria: Infantil")
elif idade < 18:
    print("Categoria: Juvenil")
elif idade < 30:
    print("Categoria: Adulto Jovem")
elif idade < 45:
    print("Categoria: Adulto Intermediário")
elif idade < 60:
    print("Categoria: Adulto Sênior")
else:
    print("Categoria: Master")
```

# Unidade 6 – Repetição Controlada
### Tabuada Personalizada com Limite
Demonstração do laço de repetição `for` associado à função `range()` para gerar uma tabuada dinâmica baseada nos limites estipulados pelo usuário.

```python
# Define o número base da tabuada e até onde ela deve ir
numero = int(input("Digite um número para ver a tabuada: "))
limite = int(input("Digite até qual número deseja calcular a tabuada: "))

print(f"\n--- Tabuada do {numero} até {limite} ---")

# O laço executa de 1 até o limite definido (incluso)
for contador in range(1, limite + 1):
    resultado = numero * contador
    print(f"{numero} x {contador} = {resultado}")
```

# Unidade 7 – Repetição Condicional
### Menu de Opções com Laço While
Construção de um menu interativo controlado por uma estrutura `while`, que continua rodando e processando opções até que o critério de parada ('Sair') seja satisfeito.

```python
# Inicializa a variável de controle do laço vazia
opcao = ""

# Executa o bloco continuamente até que o usuário digite 'Sair'
while opcao.lower() != "sair":
    print("\n--- MENU ---")
    print("1 - Opção A")
    print("2 - Opção B")
    print("3 - Opção C")
    print("Digite 'Sair' para encerrar.")

    opcao = input("Escolha uma opção: ")

    # Verifica as escolhas do usuário
    if opcao == "1":
        print("Você escolheu a Opção A")
    elif opcao == "2":
        print("Você escolheu a Opção B")
    elif opcao == "3":
        print("Você escolheu a Opção C")
    elif opcao.lower() == "sair":
        # Se for digitado 'sair' em qualquer variação de caixa, o laço quebra na próxima validação
        pass
    else:
        print("Opção inválida, tente novamente.")

print("Programa encerrado com sucesso.")
```
